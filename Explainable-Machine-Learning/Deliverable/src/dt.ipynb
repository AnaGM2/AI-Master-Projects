### 1.3. Transformación de variables categóricas

En este apartado del preprocesamiento, aplicamos una serie de transformaciones a las variables categóricas del dataset, para facilitar la construcción del modelo. Algunas de estas variables contienen un gran número de categorías, por lo que para poder tratarlas de forma más eficiente, se van a agrupar dichas categorías en un menor número de categorías más generales.

En primer lugar, definimos una función `transformar_nivel_estudios`, que utilizamos para agrupar todas las categorías relacionadas con el nivel de estudios del estudiante y de sus padres en cuatro grupos más generales: "Incompleta", "Secundaria", "Superior Técnica" y "Universitaria".

```python
def transformar_nivel_estudios(valor):
    """
    Transforma el nivel de estudios de un número entero a una de las cuatro nuevas categorías.
    """
    no_secundaria = {11, 26, 27, 38, 19, 12, 14, 15, 10, 9, 6, 29, 30, 35, 36, 37, 13, 20, 25}
    secundaria_completa = {1}
    superior_tecnico = {22, 39, 42, 18, 31, 33}
    superior_universitario = {2, 3, 4, 5, 40, 43, 44, 41}

    if valor in no_secundaria:
        return 0 # "Incompleta"
    elif valor in secundaria_completa:
        return 1 # "Secundaria"
    elif valor in superior_tecnico:
        return 2 # "Superior Técnica"
    elif valor in superior_universitario:
        return 3 # "Universitaria"
    else:
        return -1 # "Otro"

dict_study_level = {0: "Incompleta", 1: "Secundaria", 2: "Superior Técnica", 3: "Universitaria", -1: "Otro"}

def get_clase_estudios(n : int):
    return dict_study_level[n]
```

```python
qualification_features = ['Previous qualification', 'Mother\'s qualification', 'Father\'s qualification']

for feature in qualification_features:
    df[feature] = df[feature].apply(transformar_nivel_estudios)
```

Para manejar aquellas categorías con baja frecuencia, en algunas columnas categóricas se ha usado un umbral del $5\%$ para agrupar las categorías con frecuencias muy bajas en una categoría general -1 (equivale a una clase "otro"). Gracias a este paso, disminuimos el número de categorías en dichas variables.

## 2. Entrenamiento del modelo (Decision Tree)


En este apartado, entrenamos un modelo de árbol de decisión utilizando un *Pipeline* que incorpora tanto un procesamiento de las variables como la búsqueda de los mejores hiperparámetros mediante *Grid Search*.

El preprocesamiento que se aplica a las variables depende del tipo. A las variables categóricas se les aplica *One Hot Encoding*, convirtiéndolas en columnas binarias que representan las características de cada variable. Por otro lado, a las características binarias se les aplica un `FunctionTransformer` para asegurar que los datos sean procesados correctamente en el *Pipeline*.

```python
# Preprocesamiento por tipo de variable
preprocessor = ColumnTransformer(transformers=[
    ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features),
    ('scaler', FunctionTransformer(lambda x: x), numerical_features),
    ('binary', FunctionTransformer(lambda x: x if x.ndim > 1 else x[:, None]), binary_features)
])

# Aplicar SMOTE o RandomUnderSampler solo en el conjunto de entrenamiento
#sampler = RandomUnderSampler(random_state=SEED)  # O SMOTE(), SVMSMOTE(), ADASYN()

dt = DecisionTreeClassifier(random_state=SEED)


# Pipeline completo
dt_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier',dt)
])
```

La selección de hiperparámetros del modelo se realiza con un grid search.

```python
# Ejecutamos un Grid search y nos quedamos con el modelo de mejores hiperparámetros

param_grid = {
        'classifier__criterion': ['gini', 'entropy'],
        'classifier__max_depth': [3, 5, 7, 10, 15, 20, 25, 30],
        'classifier__min_samples_split': [2, 5, 10],
        'classifier__min_samples_leaf': [1, 2, 5]
    }

if not LOAD:
    grid_search = GridSearchCV(dt_model, param_grid, scoring='f1_weighted', cv=5, n_jobs=-1)
    grid_search.fit(X_train, y_train)

best_model = grid_search.best_estimator_ if not LOAD else load_model(model_path)
```

### 3.1. Explicación global inherente del árbol de decisión

Como es natural, un modelo de **árbol de decisión** es **inherentemente explicable de forma global**, ya que podemos interpretar con poco esfuerzo su comportamiento a la hora de clasificar cualquier instancia, y podemos analziar su estructura para extraer conclusiones globales sobre su método de clasificación. En general, **para explicar el esquema de decisión** de un modelo de árbol nos fijamos en 3 características:

1. **Profundidad del árbol** (o número de subdivisiones)
2. **Pureza de los nodos** (o grado de subdivisión de los nodos)
3. **Nodos cuya subdivisión da lugar a las ramas más grandes**

La profundidad del árbol refleja cómo de complejas llegan a ser las reglas de decisión que el árbol efecuta para clasificar. Por otro lado, la profundidad a la que aparecen las variables sobre el esquema del árbol nos da una idea sobre cuáles de las variables predictoras que le pasamos al modelo son realmente relevantes para la clasificación. Esto permite descartar las que no contribuyen o lo hacen en ínfima medida.

Si consideramos las otras dos características, somos capaces de cuantificar la importancia de cada variable predictora, teniendo en cuenta que una variable predictora es:

- Más importante cuanto mayor sea el número de nodos en los que aparece
- (1) - Más importante cuanto menos profundos sean los nodos en los que aparece
- (2) - Más importante cuanto menor sea el grado de impureza de los nodos en los que aparece

<br>

A continuación analizaremos gráficamente la estructura del árbol que hemos creado centrándonos en estas 3 características y tras ello proporcionaremos una explicación global del mismo.

```python
best_tree = best_model.named_steps['classifier']
```

Nuestro árbol es 7, lo cual quiere decir que para clasificar a un estudiante se aplica una reglas de decisión de hasta 7 cláusulas (que pueden ir variando), cada una de ellas asociada a un nodo (o subdivisión de rama) del árbol.

La clasificación como tal de una instancia se efectúa al llegar a los *nodos terminales* o *nodos hoja*, que son los que no subdividen la rama a la que pertencen y por tanto no dan paso a un nivel inferior de profundidad.

Adelantamos ya que el primer nodo hoja aparece en profundidad 4, lo cual nos dice que necesitaremos entre 4 y 7 cláusulas en las reglas de clasificación.

Las cláusulas de las reglas se conciben como reglas *if-then* que se ejecutan sobre el valor que tome una de las variables de entrada al modelo. Estas se agregan una tras otra conforme descendemos desde el nodo raíz por las distintas subramas del árbol, hasta llegar a un nodo hoja; cuya cláusula determina la clase etiquetada por el modelo.

<br>

Para ello es necesario tener en cuenta que (con la paleta por defecto) el color **amarillo** está asociado a la etiqueta ``Dropout``, el **verde claro** a ``Enrolled`` y el **azul oscuro** a ``Graduate``. Por otro lado, podemos observar que el grado de impureza de todos los nodos es exactamente 2, por lo que podemos despreciar esta información sobre el árbol sin pérdida de generalidad. También, vemos que las ramas (líneas en color negro que unen los nodos) se hacen más anchas cuantas más instancias de train han descendido por ellas. Aprovecharemos esto para dar explicaciones siguiendo las ramificaciones "más relevantes"

<br>

- **Profundidad 0** (*nodo raíz*): $\hspace{0.1 cm}$
El **nodo raíz** se subdivide en función de si ``Curricular units 2nd sem (approved)`` supera el umbral de $3.5$ o si no. Además, gracias al plot de las estadísticas que ofrece Sueprtree, podemos intuir ya mismo que el árbol va a clasificar la mayoría de las instancias que superen este umbral como ``Enrolled``, y la mayoría de las que no lo superen como ``Dropout``. Con esto, sabemos que el modelo, se fija en la variable ``Curricular units 2nd sem (approved)`` para arrancar su algoritmo de clasificación, por lo que intuitivamente, es la variable más importante para el modelo; ya que el resto de cláusulas para la decisión se agregan sobre esta, resultando en que esta división condiciona todas las reglas de decisión del árbol por estar asociada al nodo raíz. De ahí que la *importancia* de la variable sea muy grande en relación a las demás variables, como veremos posteriormente.

<br>

**Análisis global de ramificación 1**: $\hspace{0.1 cm}$ Cuando una instancia es clasificada por este modelo de árbol, primeramente se analiza su valor en la variable ``Curricular units 2nd sem (approved)`` en el nodo raíz, donde se realiza un *primer diagnóstico* que podemos resumir como: ***Si el estudiante ha aprobado menos de 4 asignaturas en el segundo semestre, muy probablemente va a ser un ``Dropout``, mientras que si aprueba 4 o más, es bastante probable que los finalice. Sin embargo, en este caso también podría ocurrir que los continúe o incluso que los abandone; aunque con baja verosimilitud***.

Tras esto, el árbol sigue *diagnosticando* la instancia de una forma u otra en función de si se supera o no el umbral ($3.5$) establecido sobre la ramificación previa. De todas las instancias de train, unas 2600 superan el umbral y solo unas 1300 no lo superan (es decir, en nuestra muestra hay 2/3 de los alumnos que aprueban bastante en el segundo semestre y solo 1/3 que aprueban poco). De hecho, esto muestra una correlación fuerte entre la proporción de instancias clasificadas como ``Dropout`` (aproximadamente 1/3) y las proporciones de instancias que no superan el umbral que comentamos.

Si descendemos por la rama asociada a aprobar menos de 4 asignaturas en el segundo semestre, se pasa a tener en cuenta la variable ``Curricular units 2nd sem (enrolled)`` que mide el número de asignaturas matriculadas en el seguno semestre. En este nodo, se realiza una división que diferencia entre alumnos que no se han matriculado de ninguna asignatura ese semestre y los que sí. Esta diferenciación es clave ya que no tiene sentido medir si se aprueban más o menos asignaturas en un semestre si el alumno no se ha matriculado de ninguna de ellas, y por otra parte, no es lo mismo aprobar 3 asignaturas de 3 matriculadas, que aprobar 3 matriculándose de 6. Esta diferencia conceptual (llevada al extremo) es lo que está realizando el árbol en este nodo.
En particular, si sobre lo anterior resulta que un alumno no se ha matriculado de niguna asignatura del segundo semestre, podría pertenecer a la clase ``Dropout`` o a la clase ``Graduate``con probabilidades similares, pero si resulta que el alumno si se ha matriculado de alguna asignatura ese semestre, entonces es bastante probable que abandone los estudios. De hecho, la conclusión que podemos sacar si analizamos estas dos priemras reglas en conjunto, es que ***los alumnos que aprueban pocas asignaturas del segundo semestre pero se han matriculado de varias de ellas, muy probablemente acaben abandonando los estudios***.

Si por contra descendemos desde el nodo raíz por la rama asociada a aprobar 4 o más asignaturas, encontramos un nodo que vuelve a tomar en cuenta la variable ``Curricular units 2nd sem (approved)`` para su ramificación. En este caso, el umbral de decisión sobre la variable es $5.5$, lo cual podemos interpretar que significa que si el alumno aprueba 5 o menos asignaturas del segundo semestre sería igual de verosímil que completase, continuase o abandonase los estudios, pero sin embargo, si aprueba 6 o más podemos determinar con alta confianza que el alumno finalizará sus estudios. De hecho, analizando el histograma asociado a este nodo que comentamos, podemos observar que ***cuantas más asignaturas se aprueben del segundo semestre, más verosímil es que un alumno finalice sus estudios en relación a las otras dos posibilidades (abandonar o continuar)***.

<br>

**Análisis global de ramificación 2**: $\hspace{0.1 cm}$ El siguiente paso en la clasificación de una instancia es descender a los nodos del segundo nivel de profundidad y agregar sus cláusulas asociadas a las dos que ya acumulamos  según el camino de descenso que haya recorrido la instancia (solo había dos opciones: izquierda y derecha partiendo del nodo raíz). Tenemos dos opciones de *background* hasta ahora:

1. Alumno que ha aprobado ninguna o pocas asignaturas en el segundo semestre
2. Alumno que ha aprobado varias o muchas asignaturas en el segundo semestre

Situándonos en el background 1, diferenciábamos según si se había matriculado de alguna asignatura de ese semestre o si no. Si el alumno no se ha matriculado de ninguna asignatura del segundo semestre, se evalúa agregando una nueva cláusula que toma en cuenta su nota de admisión y el umbral $136.1$ (que podemos asociar a una nota de admisión mediocre estadísticamente hablando). Solo con el background 1 es imposible de momento determinar con claridad la clase a la que pertence el alumno, así que se incorpora la variable de la nota de admisión para añadir información con la que construir un criterio de clasificación. Al añadir esta variable, lo único que averiguamos es que ***si además la nota de admisión es entre baja y mediocre, lo más probable es que el alumno abandone los estudios, pero si la nota es entre mediocre y alta, es más probable que finalice sus estudios, aunque sin mucha certeza al respecto*** ya que el umbral no separa claramente las clases en el histograma.

Si sobre el background 1 resulta que el alumno se ha matriculado de alguna asignatura del segundo semestre, entonces nos volvemos a fijar en el número de aprobados en este semestre, concretamente, en si es menor o mayor que $1.5$. Este umbral deja a la izquierda la clase ``Dropout`` mientras que a su derecha predomina ``Dropout`` pero más mezclado con las otras dos clases, lo cual significa que ***de entre los alumnos con este background que se han matriculado de asignaturas del del 2º semestre, si han aprobado 0 o 1, casi seguro abandonarán los estudios, mientras que si han aprobado 2 o 3, no está tan claro que vayan a abandonar los estudios***.

Situándonos ahora sobre el background 2, el árbol distingue entre alumnos que hayan aprobado 4 o 5 asignaturas, y alumnos con más de 5 aprobados en el segundo semestre. A los primeros se les "consulta" si han pagado las tasas de matrícula a tiempo o si no; y a los segundos se les filtra a través del número de asignaturas de las que se han evaluado en el primer semestre.
En el primer caso, la subdivisión es bastante acertada ya que casi todos los alumnos que no pagaron a tiempo las tasas resultan en un abandono de los estudios. En el segundo caso, la subdivisión mediante ``Curricular units 1st sem (evaluations)`` = $8.5$ no clarifica mucho ya que aunque hay una fuerte predominancia de la clase ``Enrolled`` a ambos lados del umbral, no termina de haber una división clara.

#### **Importancia de las variables**

Cuando hablamos de *importancia* de las variables de entrada a nuestro árbol de decisión, nos estamos refiriendo a su valor asociado por esta fórmula

$$ I(f) \hspace{0.4 cm} := \sum_{t \hspace{0.1 cm} \in  \hspace{0.1 cm} T \hspace{0.1 cm} : \hspace{0.1 cm}v(t) = f} \frac{N_t}{N} \cdot \Delta I_t $$

donde denotamos:

1. $I(f)$ es la *importancia* de la característica $f$
2. $T$ es el conjunto de nodos internos del árbol.
3. $v(t)$ es la característica $f_t$ utilizada para la división en el nodo $t$
4. $N_t$ es el número de muestras de entrenamiento que se clasifican agregando la cláusula del nodo $t$
5. $N$ es el número total de muestras en el conjunto de entrenamiento.
6. $t_L$ y $t_R$ son los nodos hijo izquierdo y derecho, respectivamente.
7. $N_{t_L}$ y $N_{t_R}$ son el número de muestras que llegan a los nodos hijo.
8. $I(t)$ es la impureza del nodo $t$, medida con Gini, entropía o reducción de varianza (según el criterio del árbol, en nuestro caso, entropía).
9. $\Delta I_t := I(t) - \left( \frac{N_{t_L}}{N_t} I(t_L) + \frac{N_{t_R}}{N_t} I(t_R) \right) \hspace{0.1 cm}$ es la reducción en la impureza en el nodo $t$.

<br>

El valor $I(f)$ retornado por esta fórmula se puede interpretar como una **medida tanto global como relativa de influencia en las predicciones para cada variable de entrada**. Así, podemos realizar comparaciones relativas en cuanto a lo decisiva que es la información de dos (o más) variables de entrada para dar lugar a la clasificación.

#### **Conclusiones**

De todo el análisis que hemos realizado, podemos destacar las siguientes conclusiones:

<br>

- Existen distintos perfiles de alumno (según sus datos académicos, demográficos y familiares) que el modelo clasifica correctamente con un alto nivel de confianza, y son los listados en esta tercera sección tras el gráfico del árbol. Sin embargo, también existen ciertos perfiles de alumno a los que el modelo no es capaz de clasificar correctamente con alto nivel de confianza, pues no consigue *separar* bien las diferentes clases a través de sus reglas de clasificación tomando en cuenta las variables de entrada seleccionadas.

<br>

- Las variables que más impacto tienen en la clasificación por parte de nuestro modelo de árbol pueden ser englobadas en distintos grupos semánticos:


1. Asignaturas aprobadas / matriculadas / evaluadas de los semestres 1 y 2
2. El Pago de tasas de matrícula a tiempo (por sí sola es importante)
3. El rendimiento académico en nota media de acceso, estudios previos o por semestre
4. El PIB, ratio de inflación y ratio de desempleo del país de origen
5. Edad y Género
6. Cuestiones relativas al grado escogido y la formación de los progenitores

En general, un número de aprobados muy bajo (menor o igual que 2) indica que el alumno abandonará los estudios, uno alto indica que el alumno se graduará y uno medio indica que seguirá matriculado, aunque también puede graduarse. Un retraso en el pago de las tasas de matrícula es un indicativo de abandono.

<br>


- Posiblemente existe un sesgo en los datos de train, porque las mujeres tienden más hacia la clase ``Graduate`` que los hombres que se reparten más entre las 3 clases, al menos para cierto perfil concreto de estudiante (similar al perfil 7 de ``Graduate``).

<br>

- En general, el modelo de árbol es más fiable cuando clasifica a un alumno como ``Dropout``, y es menos fiable al clasificar como ``Enrolled`` o ``Graduate`` ya que un gran número de las muestras de train se corresponden con perfiles de alumno en los que la regla de clasificación del árbol no separa correctamente estas dos clases.
