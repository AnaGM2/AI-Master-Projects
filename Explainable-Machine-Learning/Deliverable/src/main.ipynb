Este proyecto práctico tiene como objetivo aplicar los conceptos aprendidos en la asignatura *Machine Learning Explicable* para explicar diferentes modelos de Machine Learning mediante métodos de explicabilidad.

### **Dataset**

El dataset utilizado en este proyecto es [*Predict Students' Dropout and Academic Success*](https://archive.ics.uci.edu/dataset/697/predict+students+dropout+and+academic+success), creado por una institución portuguesa de educación superior. Este conjunto de datos contiene información sobre estudiantes matriculados en diferentes licenciaturas (como agronomía, diseño, educación, enfermería, periodismo, gestión, servicios sociales o tecnología). El objetivo principal es predecir el abandono y el éxito académico de los estudiantes mediante modelos de clasificación.

El dataset incluye datos demográficos, socioeconómicos y académicos de los estudiantes en el momento de su matrícula, así como su rendimiento académico al final del primer y segundo semestre. El problema se formula como una tarea de clasificación de tres categorías: abandono, matriculado y graduado, al final de la duración normal del curso.

### **Modelos**

En este proyecto, se han estudiado y explicado varios modelos de Machine Learning mediante diferentes métodos vistos en clase. Los modelos que se analizan son los siguientes:

 - [Decision Tree](https://colab.research.google.com/github/JMGO-coding/MLX_academic_performance/blob/main/src/dt.ipynb): Este modelo, por ser inherentemente explicable, se explora a través de sus propias características, como la estructura del árbol y la subdivisión de los nodos.

 - [SVM](https://colab.research.google.com/github/JMGO-coding/MLX_academic_performance/blob/main/src/svm.ipynb): En este caso, se utilizan técnicas como LIME y SHAP para explicar el modelo.

 - [XGBoost](https://colab.research.google.com/github/JMGO-coding/MLX_academic_performance/blob/main/src/xgboost.ipynb): También se utilizan técnicas como LIME y SHAP para explicar el modelo.
