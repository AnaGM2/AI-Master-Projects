### 1.3. Transformación de variables categóricas


En este apartado del preprocesamiento, aplicamos una serie de transformaciones a las variables categóricas del dataset, para facilitar la construcción del modelo. Algunas de estas variables contienen un gran número de categorías, por lo que para poder tratarlas de forma más eficiente, se van a agrupar dichas categorías en un menor número de categorías más generales.

En primer lugar, definimos una función `transformar_nivel_estudios`, que utilizamos para agrupar todas las categorías relacionadas con el nivel de estudios del estudiante y de sus padres en cuatro grupos más generales: "Incompleta", "Secundaria", "Superior Técnica" y "Universitaria".

```python
def transformar_nivel_estudios(valor):
    """
    Transforma el nivel de estudios de un número entero a una de las cuatro nuevas categorías.
    """
    no_secundaria = {11, 26, 27, 38, 19, 12, 14, 15, 10, 9, 6, 29, 30, 35, 36, 37, 13, 20, 25}
    secundaria_completa = {1}
    superior_tecnico = {22, 39, 42, 18, 31, 33}
    superior_universitario = {2, 3, 4, 5, 40, 43, 44, 41}

    if valor in no_secundaria:
        return 0 # "Incompleta"
    elif valor in secundaria_completa:
        return 1 # "Secundaria"
    elif valor in superior_tecnico:
        return 2 # "Superior Técnica"
    elif valor in superior_universitario:
        return 3 # "Universitaria"
    else:
        return -1 # "Otro"

dict_study_level = {0: "Incompleta", 1: "Secundaria", 2: "Superior Técnica", 3: "Universitaria", -1: "Otro"}

def get_clase_estudios(n : int):
    return dict_study_level[n]
```

```python
qualification_features = ['Previous qualification', 'Mother\'s qualification', 'Father\'s qualification']

for feature in qualification_features:
    df[feature] = df[feature].apply(transformar_nivel_estudios)
```

A continuación, como la variable `Nacionality` también contiene muchas categorías diferentes, y además la mayoría de ellas están representadas por un número muy reducido de instancias, hemos decidido cambiarla por una variable binaria llamada `foreign`. Para manejar aquellas categorías con baja frecuencia, en algunas columnas categóricas se ha usado un umbral del $5\%$ para agrupar las categorías con frecuencias muy bajas en una categoría general -1 (equivale a una clase ``otro''). Gracias a este paso, disminuimos el número de categorías en dichas variables.

```python
# 5% threshold for categorical features

threshold = 0.05

for feature in thres_features:
    value_counts = df[feature].value_counts(normalize=True)
    if value_counts.max() > threshold:
        df[feature] = df[feature].apply(lambda x: x if value_counts[x] > threshold else -1) # -1 = other
```

## 2. Entrenamiento del modelo (SVM)


En este apartado, entrenamos un modelo de árbol de decisión utilizando un *Pipeline* que incorpora tanto un procesamiento de las variables como la búsqueda de los mejores hiperparámetros mediante *Grid Search*.

El preprocesamiento que se aplica a las variables depende del tipo. A las variables categóricas se les aplica *One Hot Encoding*, convirtiéndolas en columnas binarias que representan las características de cada variable. Por otro lado, a las características binarias se les aplica un `FunctionTransformer` para asegurar que los datos sean procesados correctamente en el *Pipeline*.

```python
# Preprocesamiento por tipo de variable
preprocessor = ColumnTransformer(transformers=[
    ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features),
    ('scaler', StandardScaler(), numerical_features),
    ('binary', FunctionTransformer(lambda x: x if x.ndim > 1 else x[:, None]), binary_features)
])

# Aplicar SMOTE o RandomUnderSampler solo en el conjunto de entrenamiento
#sampler = RandomUnderSampler(random_state=SEED)  # O SMOTE(), SVMSMOTE(), ADASYN()

svm = SVC(random_state=SEED, probability=True)
    # probability=True para obtener probabilidades en la predicción, lo que facilita la explicación del modelo


# Pipeline completo
svm_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier',svm)
])
```

```python
if not LOAD:
    # Ejecutamos un Grid search y nos quedamos con el modelo de mejores hiperparámetros

    param_grid = {
        'classifier__C': [0.1, 1, 10, 100],        # Regularización
        'classifier__gamma': [0.01, 0.1, 0.5],  # Coeficiente del kernel
        'classifier__kernel': ['rbf', 'sigmoid']   # Kernel
    }

    grid_search = GridSearchCV(svm_model, param_grid, scoring='f1_weighted', cv=5, n_jobs=-1, refit=True)
    grid_search.fit(X_train, y_train)

    best_model = grid_search.best_estimator_
    save_model(best_model, model_path) # Guardamos el mejor modelo

    results = pd.DataFrame(grid_search.cv_results_)
    results.to_csv(results_path, index=False)
else:
    best_model = load_model(model_path)
    results = pd.read_csv(results_path)

y_pred = best_model.predict(X_test)
f1 = f1_score(y_test, y_pred, average='weighted')


print("Mejores parámetros: C =", best_model.named_steps['classifier'].get_params()['C'],
      ", gamma =", best_model.named_steps['classifier'].get_params()['gamma'],
      ", kernel =", best_model.named_steps['classifier'].get_params()['kernel'])
print("F1-score en test:", f1)
```

## 3. Explicación del modelo

El objetivo del proyecto es dar una explicación al modelo ML utilizado. En este caso, la tarea del modelo es la clasificación de estudiantes en tres categorías: abandono (Dropout: 0), matriculado (Enrolled: 1) y graduado (Graduated: 2). El modelo que se ha entrenado es de tipo *Support Vector Machine* (SVM).

* ¿En qué consiste la explicación del modelo ML?

Nuestro modelo de ML actúa como una función que toma un ejemplo de entrada (las características de un estudiante) y devuelve una predicción (la categoría a la que pertenece). Sin embargo, si tomamos el modelo como una caja negra, no entenderemos cómo se ha llegado a esa predicción. Entender cómo se ha llegado a la predicción, o bien, cómo los valores de las características de entrada determinan qué decisión se toma, puede ser muy útil para confiar en el modelo, tomar decisiones informadas, detectar sesgos... En esta tarea del ámbito de la educación esta interpretación del modelo nos puede ayudar tanto a entender por qué unos estudiantes abandonan como a darles una tutorización personalizada.

* ¿Qué es el modelo SVM? ¿Por qué se necesita explicarlo?

El modelo *Support Vector Machine* para clasificación, que llamamos SVC (Support Vector Classifier), se basa en encontrar el conjunto de hiperplanos que mejor separa las clases en un espacio de características de alta dimensión, transformado por un kernel. La clase predicha (o su probabilidad) se asigna según la región delimitada por los hiperplanos en la que cae el ejemplo de entrada transformado.

En el caso de un kernel no lineal, el modelo SVC no es directamente interpretable como los modelos interpretables que hemos estudiado (regresión lineal y árbol de decisión). Por tanto, lo consideraremos como una caja negra y aplicaremos técnicas de explicabilidad para entender cómo llega a sus predicciones, tanto a nivel local (para ejemplos concretos) como global (su comportamiento general).

* ¿Cómo vamos a explicar el modelo?

Los métodos de explicabilidad que vamos a aplicar son SHAP y LIME. Se trata de métodos locales, los usaremos para entender cómo el modelo ha llegado a una predicción concreta para un estudiante concreto. Gracias a distintas métricas de agregación de los valores SHAP también proporcionaremos algunas explicaciones globales del modelo. Las explicaciones globales nos darán la importancia de cada variable y para ciertas variables, también la relación entre su valor y la predicción.

A continuación en la sección 3.1. mostramos las primeras explicaciones locales por el método SHAP, en la sección 3.2. trataremos de obtener una explicación global del SVC y en la 3.3 realizaremos explicaciones locales para algunas instancias que hemos seleccionado por su interés.

### 3.1. Explicación local con SHAP

SHAP es un método de explicabilidad de modelos de ML basado en la teoría de valores de Shapley para juegos cooperativos. Una explicación local consiste en el cómputo de los valores SHAP para cada uno de los valores que toman los features de un ejemplo concreto, que son aproximaciones de los valores de Shapley. El valor de Shapley en ML mide, para un feature con su valor instanciado, cuánto contribuye ese valor del feature al valor final de la predicción. En nuestro caso la predicción es la probabilidad de pertenecer a cada una de las clases, así que a cada feature se le asocia un valor de Shapley para cada clase, que es positivo si el valor del feature en ese ejemplo incrementa la probabilidad predicha por el modelo, de la clase correspondiente, y negativo si la decrementa.

En este apartado, mostraremos las primeras explicaciones locales para ejemplos de test aleatorios. Esto nos ayudará a mostrar cómo interpretar la salida del modelo para un ejemplo concreto y a comprender la utilidad de esta explicación, sentando las bases para la interpretación de ejemplos más interesantes en la sección 3.3.

A continuación construimos el explicador SHAP (objeto `explainer`) y obtenemos las explicaciones para un ejemplo concreto. Mostramos los valores de Shapley para cada feature y cada clase, y la contribución de cada feature a la predicción final.

#### Bar plots de agregación de valores SHAP

En los siguientes gráficos mostramos para cada feature y cada clase, la suma de valores absolutos SHAP. El valor absoluto medio de los valores SHAP de una característica para una clase es una medida de la contribución media de esa característica a la predicción de la probabilidad de pertenencia a esa clase. Entonces es un indicador de la importancia de cada característica para el modelo.

```python
#@title SHAP Summary plot

target_names = ['Dropout', 'Enrolled', 'Graduate']
shap.summary_plot(shap_values, X_test[:TAM_SHAP], class_names=target_names, rng=SEED)
```

En el gráfico de barras ``mean(|SHAP value|)`` vemos, para cada característica, el valor de SHAP absoluto medio por cada clase objetivo. El color indica la clase objetivo: Enrolled (azul), Dropout (rosa) y Graduate (verde). Las variables aparecen ordenadas de arriba abajo de mayor a menor suma del valor absoluto SHAP promedio para las 3 clases objetivo (longitud de la barra).

El valor absoluto medio del valor SHAP de una característica y una clase es un indicador de la contribución media (global) de la característica para la predicción de esa clase, y la suma de los valores absolutos medios de cada clase es una medida de la contribución global de cada variable al modelo de clasificación. En definitiva, una mayor longitud de la barra indica una mayor contribución de la variable en nuestro modelo. Este valor es un indicador de la importancia de cada variable, pero no nos sirve para identificar qué tipo de relación hay entre el valor de la variable y la predicción (si es creciente, decreciente u otro). Interpretando el gráfico obtenemos las siguientes conclusiones:

- La variable ``Curricular units 2nd sem (approved)`` es la que tiene un mayor valor SHAP promedio, especialmente para la clase *Dropout*, lo que indica que es la variable más determinante para la clasificación.

- Dos variables destacan notablemente sobre el resto con un valor SHAP absoluto medio mucho mayor: ``Curricular units 2nd sem (approved)`` y ``Curricular units 1st sem (approved)``.

- Les siguen  ``Tuition fees up to date``, ``Course`` y ``Curricular units 2nd sem (enrolled)``, que tienen un valor SHAP absoluto medio algo mayor que $0.1$ entre las tres clases.

- Observamos un punto de silla en la suma de los valores SHAP promedio a partir de ``Father's occupation`` a partir del cual las variables tienen progresivamente menor valor SHAP promedio.

A continuación se desglosan gráficos de barras de los valores SHAP absolutos medios para cada una de las clases:

```python
fig, axes = plt.subplots(nrows = 2, ncols = 2, figsize=(32, 24))
for i, name in enumerate(target_names):
    ax = axes[i//2, i%2]
    plt.sca(ax)  # Asegura que los gráficos se dibujen en el eje correcto
    shap.plots.bar(shap_values[:, :, i], max_display = 15, show = False, ax=ax)
    ax.set_title(f"Valor SHAP absoluto medio de la clase {name}")

# Obtener el valor SHAP absoluto medio por característica y por clase
mean_abs_shap = np.abs(shap_values.values).mean(axis=0)

importance_order = np.argsort(np.sum(mean_abs_shap, axis = -1), axis=0)  # Índices de las características ordenadas por importancia

# Número de clases y características
n_features, _ = mean_abs_shap.shape

x = np.arange(n_features)  # Índices para cada característica

width = 0.25  # Ancho de las barras para que no se solapen

ax = axes[1, 1]

# Crear una barra para cada clase desplazándolas horizontalmente
for i, class_name in enumerate(target_names):
    ax.barh(x + i * width, mean_abs_shap[importance_order, i], height=width, label=class_name)

# Formatear el gráfico
ax.set_yticks(x + width, X_test.columns[importance_order], ha='right')
ax.set_xlabel("Valor SHAP absoluto medio")
ax.set_title("Valor SHAP absoluto medio por clase")
ax.legend(loc = 'lower right')

plt.show()
```

Llegamos a las siguientes observaciones a partir de las gráficas:

- ``Curricular units 2nd sem (approved)`` es la variable que más contribuye a la predicción del modelo para las tres clases, porque es la que tiene un valor SHAP absoluto medio más alto en los tres casos. En el caso de ``Dropout`` tiene un valor SHAP absoluto medio de $0.18$, que duplica al segundo más alto, el de ``Curricular units 1st sem (approved)``, y considerablemente mayor que el resto. Esto indica que esta característica es la más importante para el modelo para determinar si una instancia pertenece o no a la clase *Dropout*.

- ``Curricular units 1st sem (approved)`` es la segunda variable con mayor valor SHAP absoluto medio para las tres clases. Para las clases *Enrolled* y *Graduated* su valor SHAP ($0.6$ y $0.14$) es más cercano al de la primera ($0.7$ y $0.19$) que para *Dropout*. Esto indica que el número de unidades aprobadas en cada semestre es una característica muy importante para la predicción del modelo.

- Para cada clase, las siguientes características también tienen contribuciones considerables (valor SHAP absoluto medio mayor que $0.025$):
  - ``Dropout``: *Tuition fees up to date*, *Curricular units 2nd sem (enrolled)*, *Course* y *Curricular units 2nd sem (grade)*.
  - ``Enrolled``: *Course*, *Curricular units 2nd sem (evaluations)*. La segunda (el número de examenes en el segundo semestre) contribuye principalmente a que el modelo distinga si los estudiantes continuarán sus estudios o no, pues para las otras dos clases su valor SHAP absoluto medio es menor.
  - ``Graduated``: *Course*, *Curricular units 2nd sem (enrolled)*, *Curricular units 1st sem (enrolled)*, *Tuition fees up to date*, *Curricular units 2nd sem (grade)* y *Scholarship holder*. Salvo la última, estas características coinciden con las de la clase *Dropout*, así que podemos decir que estas características, que tienen que ver con el curso, los pagos y las asignaturas cursadas, ayudan a que el modelo decida entre predecir el abandono o la graduación del estudiante.

### 3.3. Explicaciones locales con LIME

Creemos que los puntos más interesantes para explicar localmente son aquellos en los que el modelo comete errores o está menos seguro de su predicción. Para realizar las explicaciones locales usaremos LIME, que es un método de explicabilidad local que se basa en construir un modelo surrogado interpretable sobre puntos cercanos a la instancia que queremos explicar. En nuestro caso usaremos un modelo surrogrado de regresión lineal, que es inherentemente explicable por los coeficientes correspondientes a cada variable. De esta manera, la ventaja que tiene sobre SHAP, que expresa la contribución de cada característica del ejemplo a la clasificación, es que los coeficientes del modelo surrogado expresan dirección: en muchos casos podremos deducir qué características del ejemplo debemos aumentar o disminuir para reducir el riesgo de abandono predicho por el modelo, o aumentar la probabilidad predicha de graduación. La desventaja que tiene es que con estos coeficientes no obtenemos la importancia de cada variable para la predicción, ni siquiera la contribución real como con SHAP.

```python
# Obtener el modelo sin preprocesamiento
best_model_final = best_model.named_steps['classifier']

# Preprocesar X_train y X_test
X_train_transformed = best_model.named_steps['preprocessor'].transform(X_train)
X_test_transformed = best_model.named_steps['preprocessor'].transform(X_test)
```

```python
# Ajustamos el preprocesador antes de intentar acceder a los nombres de las características
preprocessor = best_model.named_steps['preprocessor']

feature_names = []

# Acceder a los transformers ajustados
for name, transformer, columns in preprocessor.transformers_:
    if hasattr(transformer, 'get_feature_names_out'):
        feature_names.extend(transformer.get_feature_names_out(columns))
    else:
        # Si no tiene get_feature_names_out, usamos los nombres originales
        feature_names.extend(columns)

# Convertimos a lista de strings
feature_names = list(map(str, feature_names))
feature_index = [i for i, feat in enumerate(feature_names)]
print(feature_names)

categorical_features_names = [feat for feat in feature_names if not feat in numerical_features]
print(categorical_features_names)

categorical_features_index = [i for i, feat in enumerate(feature_names) if feat in categorical_features_names]
print(categorical_features_index)
```

El explicador LIME trabaja sobre el modelo y no sobre el pipeline, así que debemos utilizar los datasets preprocesados: tras el escalado y la conversión de variables categóricas a one hot encoding.

#### Ejemplos con baja confianza

Para realizar la clasificación, el modelo obtiene una probabilidad de salida para cada clase objetivo (*Dropout*, *Enrolled* y *Graduate*). Podemos interpretar la probabilidad de salida para la clase predicha como la confianza que tiene el modelo sobre su decisión. Tomamos los 3 ejemplos de test con menor probabilidad de salida para la clase predicha como los ejemplos de menor confianza. En estos ejemplos será interesante ver qué características influyen en aumentar o disminuir la probabilidad de cada una de las clases, y observar cómo sus características contribuyen en distintos sentidos a la predicción del modelo.

```python
probs = best_model.predict_proba(X_test)
y_pred = np.argmax(probs, axis=1)
confidence = probs.max(axis=1)  # Tomamos la probabilidad más alta de cada predicción
least_confident_idx = np.argsort(confidence)[:3]  # Seleccionamos los 3 ejemplos menos seguros
print("Índices de los ejemplos menos seguros:", least_confident_idx)
```

```python
# LIME explainer

explainer_lime = LimeTabularExplainer(X_train_transformed,
                                 feature_names = feature_names,
                                 class_names=['Dropout', 'Enrolled', 'Graduate'],
                                 mode='classification',
                                 sample_around_instance=True,
                                 discretize_continuous=False,
                                 categorical_features=categorical_features_index,
                                 #feature_selection='lasso_path', #alternativa
                                 random_state=SEED)
```

Construimos un explicador LIME para nuestro modelo SVC. Hemos usado los siguientes parámetros:
- Kernel por defecto: exponencial de ancho $0.75\sqrt{d}$, donde $d$ es el número de características del dataset.
- Muestras centradas en la instancia a explicar.
- Modelo surrogado: por defecto utiliza un modelo lineal obtenido por regresión de Ridge (regularización L2) con $\lambda=1$. La regresión de Ridge obtiene los coeficientes de la regresión lineal que minimiza la suma de los mínimos cuadrados más el coeficiente de penalización, que es proporcional en un factor $\lambda$ a la norma L2 de los coeficientes $\beta_j$:
$$\min \{SSE + \lambda \sum_{j=1}^{p}{\beta_{j}^2}\}$$

El objetivo de esta regularización es mitigar el problema de la multicolinealidad, que causa inestabilidad en los coeficientes de la regresión lineal. La regresión de Ridge reduce la varianza, mitiga el sobreajuste y también evita que demasiados pesos vayan a 0 (como ocurre con la regularización L1).

- Selección de características: como el número de características del dataset es mayor que 6, por defecto las variables seleccionadas para el modelo surrogado se obtienen por el método ``highest weights``. Este método selecciona las variables con mayor producto de valor original por peso en valor absoluto, $|\beta_j x_j|$, en una regresión de Ridge con $\lambda=0.01$ entrenada con los mismos datos que el modelo surrogado. Esto puede no reflejar el impacto real de las variables, especialmente cuando están fuertemente correlacionadas.

Los siguientes gráficos de barras muestran los pesos del modelo lineal surrogado construido por LIME para cada una de las clases objetivo y las características más importantes del ejemplo según LIME.

```python
def show_lime_local_explanation(i, explainer_lime=explainer_lime, num_features=10):
    print("True label:", inverse_target_mapping[y_test.iloc[i]])
    print("Predicted label:", inverse_target_mapping[y_pred[i]])
    local_explanation = explainer_lime.explain_instance(X_test_transformed[i].reshape(1, -1).flatten(), best_model_final.predict_proba, num_features=num_features, top_labels=3)
    local_explanation.show_in_notebook(show_all=False)
    return local_explanation

def show_class_local_explanation(local_explanation, class_index):
    local_explanation.as_pyplot_figure(label=class_index)
    plt.show()
```
